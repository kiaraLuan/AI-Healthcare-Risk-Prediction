```python
# Group the 21 DASS questions using the actual dataset column names

stress_items = [
    f'Q3_{i}_S{i}' for i in range(1, 8)
]

anxiety_items = [
    f'Q3_{i}_A{i-7}' for i in range(8, 15)
]

depression_items = [
    f'Q3_{i}_D{i-14}' for i in range(15, 22)
]

# Verify the question groups
print("Stress questions:", stress_items)
print("Anxiety questions:", anxiety_items)
print("Depression questions:", depression_items)

print("Total questions:",
      len(stress_items) + len(anxiety_items) + len(depression_items))
```

```python
print("Score columns available:")
print([col for col in df.columns if 'score' in col.lower()])
```

```python
def classify_stress(s):
    if s <= 14: return 'Normal'
    elif s <= 18: return 'Mild'
    elif s <= 25: return 'Moderate'
    elif s <= 33: return 'Severe'
    return 'Extremely Severe'

def classify_anxiety(s):
    if s <= 7: return 'Normal'
    elif s <= 9: return 'Mild'
    elif s <= 14: return 'Moderate'
    elif s <= 19: return 'Severe'
    return 'Extremely Severe'

def classify_depression(s):
    if s <= 9: return 'Normal'
    elif s <= 13: return 'Mild'
    elif s <= 20: return 'Moderate'
    elif s <= 27: return 'Severe'
    return 'Extremely Severe'

df['stress_category']     = df['stress_score'].apply(classify_stress)
df['anxiety_category']    = df['anxiety_score'].apply(classify_anxiety)
df['depression_category'] = df['depression_score'].apply(classify_depression)
```

```python
severity_rank = {'Normal': 0, 'Mild': 1, 'Moderate': 2, 'Severe': 3, 'Extremely Severe': 4}

cats = ['stress_category', 'anxiety_category', 'depression_category']
df['max_severity'] = df[cats].apply(lambda col: col.map(severity_rank)).max(axis=1)

def simplify(rank):
    if rank <= 1: return 'Low'
    elif rank <= 2: return 'Moderate'
    return 'High'

df['mental_risk_category'] = df['max_severity'].apply(simplify)
print(df['mental_risk_category'].value_counts())
print((df['mental_risk_category'].value_counts(normalize=True) * 100).round(2))
```

```python
df.to_csv('../data/processed/mental_processed.csv', index=False)

print("Processed dataset saved successfully!")
print("Dataset shape:", df.shape)
print("Target column exists:", 'mental_risk_category' in df.columns)
```
