```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# adjust the filename if yours differs
df = pd.read_csv('../data/raw/diabetes_012_health_indicators_BRFSS2015.csv')

print(df.shape)
df.info()
df.head()
```

```python
print("Missing values:\n", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
df.describe().T
```

```python
counts = df['Diabetes_012'].value_counts().sort_index()
print(counts)
print((df['Diabetes_012'].value_counts(normalize=True).sort_index() * 100).round(2))

counts.plot(kind='bar', title='Diabetes_012 class distribution')
plt.xlabel('0 = No diabetes, 1 = Prediabetes, 2 = Diabetes')
plt.show()
```

```python
plt.figure(figsize=(12, 9))
sns.heatmap(df.corr(), cmap='coolwarm', center=0)
plt.title('Correlation heatmap')
plt.show()
```

```python
sample = df.sample(n=20000, random_state=42)
sample.to_csv('../data/processed/chronic_sample.csv', index=False)
print(sample.shape)
```
